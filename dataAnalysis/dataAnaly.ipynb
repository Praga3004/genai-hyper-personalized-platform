```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
data=pd.ExcelFile('../resource/People Data Sample.xlsx')
```

```python
sheets= data.sheet_names
sheets
```

```python
for sheet in sheets:
    df=data.parse(sheet)
    print(f"Sheet Name: {sheet}")
    print(f"columns: {df.columns.tolist()}")
    print(f"Number of Instances: {df.shape[0]}")
    print("\n")
```

```python
print("Data Analysis Started")
print('\n')
print(data.sheet_names[0])

df=data.parse(data.sheet_names[0])
print(df.head())
print(f"Number of columns {df.shape[1]} Number of Instances: {df.shape[0]}")
print(f"{df.columns.tolist()}")
cols=df.columns.tolist()
for col in cols:
    print(f"Unique Values in '{col}' column: {len(set(df[col].tolist()))} Instances Values: \n{set(df[col].tolist())}")
    print('\n')
```
